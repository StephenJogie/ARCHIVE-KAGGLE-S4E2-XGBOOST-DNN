```python
def y_numbering(output:str) -> int:
    if(output == "Normal_Weight"):
        return 0
    if(output == "Insufficient_Weight"):
        return 1
    if(output == "Overweight_Level_I"):
        return 2
    if(output == "Overweight_Level_II"):
        return 3
    if(output == "Obesity_Type_I"):
        return 4
    if(output == "Obesity_Type_II"):
        return 5
    return 6

def is_arg_true(string:str,arg:str) -> bool:
    if(string==arg):
        return 1
    return 0

def yes_or_no_bool(thing:str) -> bool:
    if(thing == "yes"):
        return 1
    return 0

def male_or_female(gender:str) -> bool: 
    if(gender == "Male"):
        return 1
    return 0

def frequency_numbering(freq:str) -> int:
    if(freq =="no"):
        return 0
    if(freq == "Sometimes"):
        return 1
    if(freq == "Frequently"):
        return 2
    return 3

def y_restringing(label:int) -> str: 
    if(label==0):
        return "Normal_Weight"
    if(label==1):
        return "Insufficient_Weight"
    if(label==2):
        return "Overweight_Level_I"
    if(label==3):
        return "Overweight_Level_II"
    if(label==4):
        return "Obesity_Type_I"
    if(label==5):
        return "Obesity_Type_II"
    return "Obesity_Type_III"
```

```python
#Convert string stuff to labels 
train['usePublicT'] = train['MTRANS']=="Public_Transportation"
train['useCar'] = train['MTRANS']=="Automobile"
train['useBike'] = train['MTRANS']=="Bike"
train['useMotorbike'] = train['MTRANS']=="Motorbike"
train['useFeet'] = train['MTRANS']=="Walking"
train['CAEC'] = train['CAEC'].apply(frequency_numbering)
train['CALC'] = train['CALC'].apply(frequency_numbering)
train['family_history_with_overweight'] = train['family_history_with_overweight'].apply(yes_or_no_bool)
train['FAVC'] = train['FAVC'].apply(yes_or_no_bool)
train['SMOKE'] = train['SMOKE'].apply(yes_or_no_bool)
train['SCC'] = train['SCC'].apply(yes_or_no_bool)
train['Gender'] = train['Gender'].apply(male_or_female)
train['NObeyesdad'] = train['NObeyesdad'].apply(y_numbering)
train.drop('MTRANS',inplace=True,axis=1)
train.drop('id',inplace=True,axis=1)
#print(train['usePublicT'])
```

```python
test['usePublicT'] = test['MTRANS']=="Public_Transportation"
test['useCar'] = test['MTRANS']=="Automobile"
test['useBike'] = test['MTRANS']=="Bike"
test['useMotorbike'] = test['MTRANS']=="Motorbike"
test['useFeet'] = test['MTRANS']=="Walking"
test['CAEC'] = test['CAEC'].apply(frequency_numbering)
test['CALC'] = test['CALC'].apply(frequency_numbering)
test['family_history_with_overweight'] = test['family_history_with_overweight'].apply(yes_or_no_bool)
test['FAVC'] = test['FAVC'].apply(yes_or_no_bool)
test['SMOKE'] = test['SMOKE'].apply(yes_or_no_bool)
test['SCC'] = test['SCC'].apply(yes_or_no_bool)
test['Gender'] = test['Gender'].apply(male_or_female)
test.drop('MTRANS',inplace=True,axis=1)
test.drop('id',inplace=True,axis=1)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.utils import shuffle
import tensorflow as tf
Y = train['NObeyesdad']
X = train.copy()

X.drop('NObeyesdad',inplace=True,axis=1)
X = np.asarray(X).astype(np.float32)
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2)
X_train, Y_train = shuffle(X_train, Y_train)
X_test, Y_test = shuffle(X_test, Y_test)
train_y_categorical = tf.keras.utils.to_categorical(Y_train, num_classes=7)
test_y_categorical = tf.keras.utils.to_categorical(Y_test, num_classes=7)
```

```python
#tf.config.set_visible_devices([], 'GPU')
#tf.config.list_physical_devices
print(tf.test.is_gpu_available())
#tf.config.list_physical_devices('GPU')
```

```python
model = tf.keras.Sequential()
#model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.Dense(256, activation=tf.nn.relu, input_shape = (20,)))
model.add(tf.keras.layers.Normalization(axis=-1))
#model.add(tf.keras.layers.Dropout(0.5))
model.add(tf.keras.layers.Dense(128, activation=tf.nn.relu))
#model.add(tf.keras.layers.Dropout(0.5))
#model.add(tf.keras.layers.Dense(256, activation=tf.nn.relu))
#model.add(tf.keras.layers.Dense(128, activation=tf.nn.relu))
#model.add(tf.keras.layers.Dropout(0.5))
model.add(tf.keras.layers.Dense(64, activation=tf.nn.relu))
model.add(tf.keras.layers.Dense(7, activation=tf.nn.softmax))

#tf.keras.optimizers.Optimizer
#optimizer = tf.keras.optimizers.Adam(learning_rate=0.0001)
#optimizer = tf.keras.optimizers.SGD(learning_rate=0.01)
model.summary()
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
history = model.fit(X_train,train_y_categorical, epochs=50, validation_data=(X_test ,test_y_categorical))
```

```python
sample_solution['NObeyesdad'] = model.predict(test)
sample_solution['NObeyesdad'] = sample_solution['NObeyesdad'].apply(y_restringing)
sample_solution.to_csv('s4e2_attempt_2_dnn.csv', index=False)
```
